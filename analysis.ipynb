```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(10,6))
sns.set(style="whitegrid")
```

```python
df = pd.read_csv("demand_forecasting.csv")
df.head()
```

```python
df.dtypes
```

```python
df["Date"] = pd.to_datetime(df["Date"])
```

```python
df.dtypes
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
df.duplicated().sum()
```

```python
df.describe().T
```

```python
df.describe(include="object").T
```

```python
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Day"] = df["Date"].dt.day
df["Weekday"] = df["Date"].dt.day_name()
```

```python
df.head()
```

```python
df.columns
```

```python
df["discounted Price"] = df["Price"] * (1 - df["Discount"] / 100)
```

```python
df.head()
```

```python
df["Sell Through Rate"] = df["Units Sold"] / df["Inventory Level"]
```

```python
df.head()
```

```python
df.groupby("Category")["Demand"].agg(["mean","sum","std"]).sort_values(by="sum", ascending=False).reset_index()
```

```python
df.groupby(["Region", "Seasonality"])["Demand"].mean()
```

```python
df.groupby("Promotion")["Demand"].mean()
```

```python
pd.pivot_table(df, values = "Demand", index = "Month", columns = "Category", aggfunc = "mean")
```

```python
sns.histplot(df["Demand"], bins=20, kde=True)
plt.title("Demand Distribution")
plt.show()
```

```python
sns.scatterplot(df, x="Inventory Level", y="Units Sold")
plt.title("Inventory vs Units Sold")
plt.show()
```

```python
sns.boxplot(df, x="Category", y="Demand", hue="Category")
plt.title("Demand by Category")
plt.xticks(rotation=45)
plt.show()
```

```python
sns.boxplot(df, x="Weather Condition", y="Demand", hue="Weather Condition")
plt.xticks(rotation=45)
plt.title("Demand by Weather Condition")
plt.show()
```

```python
monthly_demand = df.groupby("Month")["Demand"].mean()
monthly_demand
```

```python
monthly_demand.plot(kind="bar")
plt.title("Average Demand by Month")
plt.show()
```

```python
daily_demand = df.groupby("Date")["Demand"].sum()
daily_demand
```

```python
daily_demand.plot()
plt.title("Toal Daily Demand Over Time")
plt.xlabel("Date")
plt.ylabel("Demand")
plt.show()
```

```python
sns.barplot(df, x="Promotion", y="Demand")
plt.title("Promotion Impact on Demand")
plt.show()
```

```python
df.columns
```

```python
sns.scatterplot(df, x="discounted Price", y="Demand")
plt.title("Discounted Price vs Demand")
plt.show()
```

```python
df.groupby("Seasonality")["Demand"].mean().plot(kind="bar", title="Demand by Seasonality")
plt.show()
```

```python
df.groupby("Epidemic")["Demand"].mean().plot(kind="bar", title="Epidemic Impact on Demand")
plt.show()
```
